```python
# Configure ADLS access through the workspace before running this notebook.
storage_account_name = "YOUR_STORAGE_ACCOUNT"
```

```python
clean_reviews_path = f"abfss://processed@{storage_account_name}.dfs.core.windows.net/clean_reviews/"
clean_reviews_df = spark.read.parquet(clean_reviews_path)
```

```python
metadata_path = f"abfss://raw@{storage_account_name}.dfs.core.windows.net/meta_Electronics_fixed.json"
metadata_df = spark.read.json(metadata_path)
```

```python
metadata_df = metadata_df.select(
"asin",
"title",
"brand",
"price"
)
```

```python
enriched_df = clean_reviews_df.join(
metadata_df,
on="asin",
how="left"
)
```

```python
enriched_reviews_path = f"abfss://processed@{storage_account_name}.dfs.core.windows.net/enriched_reviews/"
enriched_df.write.mode("overwrite").parquet(enriched_reviews_path)
```
